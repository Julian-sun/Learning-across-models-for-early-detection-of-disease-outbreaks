```python
#!/usr/bin/env python
# coding: utf-8

# =====================================================
# Descriptive analysis of PHC encounters and ILI visits
# =====================================================

# -------------------------
# 1. Import libraries
# -------------------------

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path
from datetime import datetime

import descri_function as desc_fun

# -------------------------
# 2. Load dataset
# -------------------------

data_path = Path(
    "/opt/storage/shared/aesop/aesop_shared/ensamble_modelling/aesop_2026_01_21_mun.parquet"
)

df = pd.read_parquet(data_path)


# -------------------------
# 3. Select variables
# -------------------------

variables = [
    "co_ibge",
    "epiyear",
    "epiweek",
    "atend_totais",
    "atend_ivas",
    "year_week",
    "epidemi_cal_start",
    "epidemi_cal_end",
    "year_week_ts",
    "co_uf",
    "nm_uf",
    "co_ibge7",
    "nm_municipio",
    "sigla_uf",
]

dta = df[variables].copy()


# -------------------------
# 4. Select study period
# -------------------------

dta = dta[(dta.year_week >= "2022-42") & (dta.year_week <= "2025-32")]


# -------------------------
# 5. Total encounters
# -------------------------

total_ili = dta["atend_ivas"].sum()
total_encounters = dta["atend_totais"].sum()

ili_proportion = round(total_ili * 100 / total_encounters, 2)

print("Total PHC encounters:", total_encounters)
print("Total ILI encounters:", total_ili)
print("ILI proportion (%):", ili_proportion)


# -------------------------
# 6. Weekly aggregated data
# -------------------------

weekly = (
    dta.groupby("year_week")[["atend_ivas", "atend_totais"]]
    .sum()
    .reset_index()
)

weekly["prop_atend"] = round(
    weekly["atend_ivas"] * 100 / weekly["atend_totais"], 2
)


# --------------------------------------------
# 7. Descriptive statistics weekly aggregation
# --------------------------------------------

# Mean weekly ILI encounters with 95% CI
print('Mean weekly ILI encounters with 95% CI', desc_fun.mean_confidence_interval(
    weekly["atend_ivas"], confidence=0.95
))

# Mean weekly total encounters with 95% CI
print('Mean weekly total encounters with 95% CI',desc_fun.mean_confidence_interval(
    weekly["atend_totais"], confidence=0.95
))

# Mean weekly proportion of ILI encounters with 95% CI
print('Mean weekly proportion of ILI encounters with 95% CI',desc_fun.mean_confidence_interval2(
    weekly["prop_atend"], confidence=0.95
))

# -------------------------
# 8. city aggregated data
# -------------------------

city = (
    dta.groupby("co_ibge")[["atend_ivas", "atend_totais"]]
    .sum()
    .reset_index()
)

city["prop_atend"] = round(
    city["atend_ivas"] * 100 / city["atend_totais"], 2
)

# -------------------------------------------
# 9. Descriptive statistics city aggregation
# -------------------------------------------

# Mean weekly ILI encounters with 95% CI
print('Mean city ILI encounters with 95% CI', desc_fun.mean_confidence_interval(
    city["atend_ivas"], confidence=0.95
))

# Mean weekly total encounters with 95% CI
print('Mean city total encounters with 95% CI',desc_fun.mean_confidence_interval(
    city["atend_totais"], confidence=0.95
))

# Mean weekly proportion of ILI encounters with 95% CI
print('Mean city proportion of ILI encounters with 95% CI',desc_fun.mean_confidence_interval2(
    city["prop_atend"].fillna(0), confidence=0.95
))
```
